### Тапсырма 1. Ауа райы кестесі

Синоптик температураны мәтін түрінде жіберді: `start_text = "-10"`, `end_text = "40"`, `step_text = "10"`. Цельсий бойынша `start`-тан `end`-ке дейін (қоса) `step` қадаммен әр температураны Фаренгейтке айналдырыңыз. Формула: F = C * 9 / 5 + 32.

Әр температураға сипаттама беріңіз: 0-ден төмен — "аяз", 0–15 — "салқын", 16–25 — "жылы", 25-тен жоғары — "ыстық". Соңында "ыстық" санатындағы температуралар санын шығарыңыз.

**Көмекші қадамдар:**

- мәтіндік мәндерді `int()` арқылы санға айналдырыңыз;
- `range()` функциясында соңғы мәнді қосу үшін `end + 1` жазыңыз;
- сипаттаманы `if-elif-else` арқылы анықтаңыз;
- ыстық күндер санын жинақтаушы айнымалыда есептеңіз.

**Күтілетін нәтиже:**

```
 -10 C =  14.0 F - аяз
   0 C =  32.0 F - салқын
  10 C =  50.0 F - салқын
  20 C =  68.0 F - жылы
  30 C =  86.0 F - ыстық
  40 C = 104.0 F - ыстық
Ыстық температуралар саны: 2
```

````python
# Шешіміңізді осында жазыңыз
start_text = "-10"
end_text = "40"
step_text = "10"

start =int(start_text)
end =int(end_text)
step_text= int(step_text)
hot_count = 0

for c in range(start, end + 1, step_text):
    f = c * 9 / 5 + 32

    if c < 0:
      label = "aiaz"
    elif c <= 15:
      label = "salqyn"
    elif c <= 25:
      label = "zyly"
    else:
      label = "ystiq"
      hot_count+=1

    print(f"{c} C = {f:.1f} F - {label}")

print(hot_count)
````

### Тапсырма 2. Жинақ шоты

Банк қосымшасынан келген деректер: бастапқы салым `deposit_text = "50000"`, айлық пайыз мөлшерлемесі `rate_text = "1.5"`, мерзімі `months_text = "12"` ай. Әр айдың соңында балансқа сол айдағы баланстың 1.5 пайызы қосылады.

Әр айдың соңындағы балансты екі ондық таңбамен шығарыңыз. Баланс алғаш рет `55000` теңгеге жеткен немесе одан асқан айды анықтаңыз. Соңында жалпы табыс сомасын (соңғы баланс минус бастапқы салым) шығарыңыз.

**Көмекші қадамдар:**

- салым мен пайызды `float()`, ай санын `int()` арқылы түрлендіріңіз;
- пайызды бөлшекке айналдыру үшін оны 100-ге бөліңіз;
- мақсатқа жеткен айды сақтау үшін бастапқы мәні 0 болатын айнымалы қолданыңыз;
- бұл айнымалыны тек ол әлі 0 болғанда ғана өзгертіңіз.

**Күтілетін нәтиже:**

```
 1-ай: 50750.00 теңге
 2-ай: 51511.25 теңге
```

````python
# Шешіміңізді осында жазыңыз
deposit_text = "50000"
rate_text = "1.5"
months_text = "12"
goal = 55000

balance= int(deposit_text)
rate = float(rate_text) / 100
months = int(months_text)
start_balance = balance

goal_month =0

for month in range(1, months+1):
  balance = balance+balance*rate
  print(balance)

  if balance >=goal and goal_month ==0:
    goal_month = month

print(goal_month)
print(balance-start_balance)
````

### Тапсырма 3. Кібісе жылдар және жас

Студенттің туған жылы мәтін түрінде берілген: `birth_text = "2003"`, ағымдағы жыл `current_year = 2026`. Туған жылдан бастап ағымдағы жылға дейінгі (қоса) барлық кібісе жылдарды және сол жылы студенттің неше жаста болғанын шығарыңыз.

Жыл 4-ке бөлінсе және 100-ге бөлінбесе, немесе 400-ге бөлінсе, ол кібісе жыл болады. Соңында кібісе жылдардың жалпы санын, студенттің ағымдағы жасын және өмір сүрген күндерінің шамамен санын (жас * 365 + кібісе жылдар саны) шығарыңыз.

**Көмекші қадамдар:**

- туған жылды `int()` арқылы санға айналдырыңыз;
- кібісе жыл шартын `and`, `or` операторларымен бір шартқа жазыңыз;
- кібісе жылдар санын цикл ішінде есептеңіз;
- жасты цикл айнымалысы мен туған жылдың айырмасы арқылы табыңыз.

**Күтілетін нәтиже:**

```
2004 жыл - 1 жаста
2008 жыл - 5 жаста
2012 жыл - 9 жаста
2016 жыл - 13 жаста
2020 жыл - 17 жаста
2024 жыл - 21 жаста
Кібісе жылдар саны: 6
Ағымдағы жасы: 23
Өмір сүрген күндері (шамамен): 8401
```

````python
birth_text = "2003"
current_year = 2026

birth_year = int(birth_text)
leap_count = 0

for year in range(birth_year, current_year + 1):
    if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
        age = year - birth_year
        print(f"{year} жыл - {age} жаста")
        leap_count += 1

current_age = current_year - birth_year
days = current_age * 365 + leap_count

print("Кібісе жылдар саны:", leap_count)
print("Ағымдағы жасы:", current_age)
print("Өмір сүрген күндері (шамамен):", days)
````

### Тапсырма 4. Аты-жөнді ретке келтіру

Тіркеу формасына студент аты-жөнін қате енгізген: `raw = "   aiGERim   sadyKOVA  "`. Артық бос орындарды алып тастап, атын және тегін бөлек айнымалыларға алыңыз. Әрқайсысының бірінші әрпі бас әріп, қалғандары кіші әріп болуы керек.

Толық аты-жөнді, инициалдарды (мысалы, "A. S.") және аты-жөндегі дауысты әріптер (a, e, i, o, u) санын шығарыңыз. Егер толық аты-жөннің ұзындығы (бос орынсыз) 15 символдан асса, "Ұзын аты-жөн", әйтпесе "Қысқа аты-жөн" деп шығарыңыз. `split()` әдісін қолданбаңыз.

**Көмекші қадамдар:**

- `strip()` арқылы басы мен соңындағы бос орындарды тазалаңыз;
- бірінші бос орынның индексін `find(" ")` арқылы табыңыз;
- атын және тегін кесінді алу арқылы бөліңіз, тектің алдындағы артық бос орындарды тағы `strip()` арқылы алыңыз;
- бас әріпті `[0].upper()`, қалған бөлікті `[1:].lower()` арқылы жасаңыз;
- дауысты әріптерді `for` циклі мен `in` операторы арқылы санаңыз.

**Күтілетін нәтиже:**

```
Толық аты-жөні: Aigerim Sadykova
Инициалдары: A. S.
Дауысты әріптер саны: 7
Ұзындығы (бос орынсыз): 15
Қысқа аты-жөн
```

````python
raw = " aiGERim sadyKOVA "

# Артық бос орындарды алып тастау
raw = raw.strip()

# Аты мен тегін бөлу (split қолданбаймыз)
space_index = raw.find(" ")

first_name = raw[:space_index].strip()
last_name = raw[space_index:].strip()

# Аты-жөнді дұрыс форматтау
first_name = first_name[0].upper() + first_name[1:].lower()
last_name = last_name[0].upper() + last_name[1:].lower()

# Толық аты-жөні
full_name = first_name + " " + last_name

# Инициалдар
initials = first_name[0] + ". " + last_name[0] + "."

# Дауысты әріптерді санау
vowels = "aeiou"
vowel_count = 0

for letter in full_name.lower():
    if letter in vowels:
        vowel_count += 1

# Ұзындығын есептеу
length = len(first_name) + len(last_name)

print("Толық аты-жөні:", full_name)
print("Инициалдары:", initials)
print("Дауысты әріптер саны:", vowel_count)
print("Ұзындығы (бос орынсыз):", length)

if length > 15:
    print("Ұзын аты-жөн")
else:
    print("Қысқа аты-жөн")
````

### Тапсырма 5. Санның цифрларын талдау

Студенттің билет нөмірі мәтін түрінде берілген: `number_text = "5829174"`. Цикл арқылы нөмірдің әр цифрын қарап шығып, келесілерді табыңыз: цифрлардың қосындысы, жұп цифрлар саны, ең үлкен цифр.

Нөмірді кері ретпен шығарыңыз. Цифрлардың қосындысы 3-ке бөлінсе "Нөмір 3-ке бөлінеді", әйтпесе "Нөмір 3-ке бөлінбейді" деп шығарыңыз. Сондай-ақ нөмірдің бүтін сан ретінде жұп немесе тақ екенін анықтаңыз.

**Көмекші қадамдар:**

- жолдың әр символын `for ch in number_text` арқылы алыңыз;
- әр символды `int()` арқылы цифрға айналдырыңыз;
- ең үлкен цифрды табу үшін бастапқы мәні 0 болатын айнымалы құрып, цикл ішінде салыстырыңыз;
- кері ретті кесінді алу `[::-1]` арқылы алыңыз.

**Күтілетін нәтиже:**

```
Цифрлар қосындысы: 36
Жұп цифрлар саны: 3
Ең үлкен цифр: 9
Кері ретпен: 4719285
Нөмір 3-ке бөлінеді
Нөмір жұп
```

````python
number_text = "5829174"

digit_sum = 0
even_count = 0
max_digit = 0

# Әр цифрды тексеру
for ch in number_text:
    digit = int(ch)

    digit_sum += digit

    if digit % 2 == 0:
        even_count += 1

    if digit > max_digit:
        max_digit = digit

# Кері ретпен шығару
reverse_number = number_text[::-1]

print("Цифрлар қосындысы:", digit_sum)
print("Жұп цифрлар саны:", even_count)
print("Ең үлкен цифр:", max_digit)
print("Кері ретпен:", reverse_number)

# 3-ке бөлінуін тексеру
if digit_sum % 3 == 0:
    print("Нөмір 3-ке бөлінеді")
else:
    print("Нөмір 3-ке бөлінбейді")

# Жұп немесе тақ екенін анықтау
if int(number_text) % 2 == 0:
    print("Нөмір жұп")
else:
    print("Нөмір тақ")
````

### Тапсырма 6. Көтерме сауда және жеңілдік

Дүкенде `product = "Notebook"` тауарының бір данасының бағасы `price_text = "1250"` теңге. Сатып алу саны 1-ден `max_qty_text = "10"` данаға дейін болуы мүмкін. Жеңілдік ережесі: 5–7 дана алғанда 5 пайыз, 8 және одан көп дана алғанда 10 пайыз жеңілдік беріледі.

Әр сан үшін жеңілдікке дейінгі құнды, жеңілдік пайызын және төленетін соманы кесте түрінде шығарыңыз. Студенттің бюджеті `budget = 9000` теңге. Осы бюджетке сатып алуға болатын ең көп дана санын анықтаңыз.

**Көмекші қадамдар:**

- бағаны және ең көп санды `int()` арқылы түрлендіріңіз;
- жеңілдік пайызын `if-elif-else` арқылы анықтаңыз;
- төленетін сома: құн минус құнның жеңілдік пайызы;
- бюджетке сыятын санды цикл ішінде шартпен жаңартып отырыңыз.

**Күтілетін нәтиже:**

```
Тауар: Notebook, бағасы: 1250 теңге
Саны | Құны  | Жеңілдік | Төлем
   1 |  1250 |       0% | 1250.0
   2 |  2500 |       0% | 2500.0
   3 |  3750 |       0% | 3750.0
   4 |  5000 |       0% | 5000.0
   5 |  6250 |       5% | 5937.5
   6 |  7500 |       5% | 7125.0
   7 |  8750 |       5% | 8312.5
   8 | 10000 |      10% | 9000.0
   9 | 11250 |      10% | 10125.0
  10 | 12500 |      10% | 11250.0
Бюджетке сыятын ең көп дана: 8
```

````python
product = "Notebook"
price_text = "1250"
max_qty_text = "10"
budget = 9000

price = int(price_text)
max_qty = int(max_qty_text)

print(f"Тауар: {product}, бағасы: {price} теңге")
print("Саны | Құны  | Жеңілдік | Төлем")

max_affordable = 0

for qty in range(1, max_qty + 1):
    cost = price * qty

    if qty >= 8:
        discount = 10
    elif qty >= 5:
        discount = 5
    else:
        discount = 0

    payment = cost - cost * discount / 100

    print(f"{qty:4} | {cost:5} | {discount:7}% | {payment}")

    if payment <= budget:
        max_affordable = qty

print(f"Бюджетке сыятын ең көп дана: {max_affordable}")
````

### Тапсырма 7. Құпиясөздің сенімділігі

Пайдаланушы құпиясөз ойлап тапты: `password = "Python2026kz"`. Цикл арқылы құпиясөздегі бас әріптер, кіші әріптер, цифрлар және басқа символдар санын есептеңіз.

Төрт талапты тексеріңіз: ұзындығы кемінде 8 символ, кемінде бір бас әріп, кемінде бір кіші әріп, кемінде бір цифр. Әр талаптың орындалғанын немесе орындалмағанын шығарыңыз. Орындалған талаптар саны 4 болса "Күшті", 3 болса "Орташа", одан аз болса "Әлсіз" деп бағалаңыз. Бағдарламаны `"abc123"` құпиясөзімен де тексеріп көріңіз.

**Көмекші қадамдар:**

- символ түрін `isupper()`, `islower()`, `isdigit()` әдістерімен анықтаңыз;
- әр түр үшін бөлек санағыш айнымалы құрыңыз;
- орындалған талаптар санын жеке айнымалыда жинақтаңыз;
- соңғы бағаны `if-elif-else` арқылы қойыңыз.

**Күтілетін нәтиже:**

```
Бас әріптер: 1
Кіші әріптер: 7
Цифрлар: 4
Басқа символдар: 0
Ұзындығы: орындалды
Бас әріп: орындалды
Кіші әріп: орындалды
Цифр: орындалды
Бағасы: Күшті
```

````python

password = "Python2026kz"

# Санағыш айнымалылар
upper_count = 0
lower_count = 0
digit_count = 0
other_count = 0

# Құпиясөз символдарын санау
for char in password:
    if char.isupper():
        upper_count += 1
    elif char.islower():
        lower_count += 1
    elif char.isdigit():
        digit_count += 1
    else:
        other_count += 1

print("Бас әріптер:", upper_count)
print("Кіші әріптер:", lower_count)
print("Цифрлар:", digit_count)
print("Басқа символдар:", other_count)

# Талаптарды тексеру
length_ok = len(password) >= 8
upper_ok = upper_count >= 1
lower_ok = lower_count >= 1
digit_ok = digit_count >= 1

print("Ұзындығы:", "орындалды" if length_ok else "орындалмады")
print("Бас әріп:", "орындалды" if upper_ok else "орындалмады")
print("Кіші әріп:", "орындалды" if lower_ok else "орындалмады")
print("Цифр:", "орындалды" if digit_ok else "орындалмады")

# Орындалған талаптарды санау
requirements = 0

if length_ok:
    requirements += 1
if upper_ok:
    requirements += 1
if lower_ok:
    requirements += 1
if digit_ok:
    requirements += 1

# Құпиясөзге баға беру
if requirements == 4:
    result = "Күшті"
elif requirements == 3:
    result = "Орташа"
else:
    result = "Әлсіз"

print("Бағасы:", result)
````

### Тапсырма 8. Сөйлемді талдау

Мәтін берілген: `sentence = "Python is a simple and powerful programming language"`. Цикл арқылы мәтіндегі әріптер санын, бос орындар санын және дауысты әріптер (a, e, i, o, u) санын есептеңіз. Сөздер арасында бір ғана бос орын бар деп есептеп, сөздер санын табыңыз.

`split()` әдісін қолданбай, мәтіндегі ең ұзын сөзді және оның ұзындығын анықтаңыз. Соңында әріптердің жалпы ішіндегі дауысты әріптердің үлесін пайызбен (бір ондық таңбамен) шығарыңыз.

**Көмекші қадамдар:**

- әріпті `isalpha()` әдісімен анықтаңыз;
- сөздер саны бос орындар санынан бір артық болады;
- ағымдағы сөзді бос жолдан бастап, әр әріпті оған қосып отырыңыз;
- бос орынға жеткенде ағымдағы сөзді ең ұзын сөзбен салыстырып, оны тазалаңыз;
- соңғы сөзді де тексеру үшін мәтіннің соңына бір бос орын қосып цикл жүргізіңіз.

**Күтілетін нәтиже:**

```
Әріптер саны: 45
Бос орындар саны: 7
Дауысты әріптер саны: 16
Сөздер саны: 8
Ең ұзын сөз: programming - 11 әріп
Дауысты әріптердің үлесі: 35.6%
```

````python
sentence = "Python is a simple and powerful programming language"

# Санағыш айнымалылар
letter_count = 0
space_count = 0
vowel_count = 0

vowels = "aeiou"

# Ең ұзын сөзді анықтау
longest_word = ""
current_word = ""

# Мәтінді цикл арқылы талдау
for char in sentence:
    if char.isalpha():
        letter_count += 1

        if char.lower() in vowels:
            vowel_count += 1

        current_word += char

    elif char == " ":
        space_count += 1

        if len(current_word) > len(longest_word):
            longest_word = current_word

        current_word = ""

# Соңғы сөзді тексеру
if len(current_word) > len(longest_word):
    longest_word = current_word

# Сөздер санын анықтау
word_count = space_count + 1

# Дауысты әріптердің пайыздық үлесі
vowel_percent = vowel_count / letter_count * 100

# Нәтижелерді шығару
print("Әріптер саны:", letter_count)
print("Бос орындар саны:", space_count)
print("Дауысты әріптер саны:", vowel_count)
print("Сөздер саны:", word_count)
print(f"Ең ұзын сөз: {longest_word} - {len(longest_word)} әріп")
print(f"Дауысты әріптердің үлесі: {vowel_percent:.1f}%")
````

### Тапсырма 9. Компьютер санды табады

Студент 1-ден 100-ге дейінгі санды ойлады: `secret_text = "73"`. Компьютер бұл санды "екіге бөлу" тәсілімен табуы керек: ол әрқашан мүмкін аралықтың ортасындағы санды атайды. Егер аталған сан ойланған саннан кіші болса, аралықтың төменгі шекарасы, ал үлкен болса, жоғарғы шекарасы өзгереді.

`while` циклі арқылы компьютердің әр болжамын, оның нәтижесін ("кіші", "үлкен" немесе "табылды") шығарыңыз. Сан табылғанда цикл тоқтап, әрекеттер саны көрсетілуі керек. Әрекеттер саны 7-ден аспаса "Тиімді іздеу" деп шығарыңыз.

**Көмекші қадамдар:**

- аралық шекараларын `low = 1`, `high = 100` айнымалыларында сақтаңыз;
- ортаңғы санды `(low + high) // 2` арқылы есептеңіз;
- болжам кіші болса `low = guess + 1`, үлкен болса `high = guess - 1`;
- сан табылғанда `break` арқылы циклден шығыңыз.

**Күтілетін нәтиже:**

```
1-әрекет: 50 - кіші
2-әрекет: 75 - үлкен
3-әрекет: 62 - кіші
4-әрекет: 68 - кіші
5-әрекет: 71 - кіші
6-әрекет: 73 - табылды
Әрекеттер саны: 6
Тиімді іздеу
```

````python
secret_text = "73"
secret = int(secret_text)

low = 1
high = 100
attempts = 0

while low <= high:
    guess = (low + high) // 2
    attempts += 1

    if guess < secret:
        print(f"{attempts}-әрекет: {guess} - кіші")
        low = guess + 1

    elif guess > secret:
        print(f"{attempts}-әрекет: {guess} - үлкен")
        high = guess - 1

    else:
        print(f"{attempts}-әрекет: {guess} - табылды")
        break

print("Әрекеттер саны:", attempts)

if attempts <= 7:
    print("Тиімді іздеу")
````

### Тапсырма 10. Емтихан баллдарын талдау

Оқытушы студенттің баллдарын бір жол етіп жіберді: `scores_text = "85,92,78,64,90,55"`. `split()` әдісін қолданбай, цикл арқылы жолдан әр баллды бөліп алып, бүтін санға айналдырыңыз.

Баллдар санын, қосындысын, орташа мәнін (бір ондық таңбамен), ең жоғары және ең төменгі баллды табыңыз. 60 баллдан төмен баллдар санын анықтаңыз. Орташа балл бойынша бағаны қойыңыз: 90 және жоғары — "A", 75–89 — "B", 60–74 — "C", 60-тан төмен — "F". 60-тан төмен балл бар болса "Қайта тапсыру қажет", әйтпесе "Барлық емтихан тапсырылды" деп шығарыңыз.

**Көмекші қадамдар:**

- цифрларды ағымдағы мәтіндік айнымалыға жинап, үтірге жеткенде оны `int()` арқылы санға айналдырыңыз;
- соңғы баллды өңдеу үшін жолдың соңына бір үтір қосып цикл жүргізіңіз;
- ең жоғары балл үшін бастапқы мән 0, ең төменгі балл үшін 100 алыңыз;
- орташа мәнді қосындыны баллдар санына бөлу арқылы табыңыз.

**Күтілетін нәтиже:**

```
Баллдар саны: 6
Қосындысы: 464
Орташа балл: 77.3
Ең жоғары балл: 92
Ең төменгі балл: 55
60-тан төмен баллдар: 1
Бағасы: B
Қайта тапсыру қажет
```

````python
scores_text = "85,92,78,64,90,55"

# Бастапқы айнымалылар
current_score = ""
count = 0
total = 0
max_score = 0
min_score = 100
below_60 = 0

# split() қолданбай баллдарды бөліп алу
for char in scores_text + ",":
    if char != ",":
        current_score += char
    else:
        score = int(current_score)

        count += 1
        total += score

        if score > max_score:
            max_score = score

        if score < min_score:
            min_score = score

        if score < 60:
            below_60 += 1

        current_score = ""

# Орташа баллды есептеу
average = total / count

# Орташа балл бойынша баға қою
if average >= 90:
    grade = "A"
elif average >= 75:
    grade = "B"
elif average >= 60:
    grade = "C"
else:
    grade = "F"

# Нәтижелерді шығару
print("Баллдар саны:", count)
print("Қосындысы:", total)
print(f"Орташа балл: {average:.1f}")
print("Ең жоғары балл:", max_score)
print("Ең төменгі балл:", min_score)
print("60-тан төмен баллдар:", below_60)
print("Бағасы:", grade)

if below_60 > 0:
    print("Қайта тапсыру қажет")
else:
    print("Барлық емтихан тапсырылды")
````
